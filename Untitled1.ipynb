```python
import torch
import torch.nn as nn
import torch.nn.functional as F

from torch.utils.data import Dataset, DataLoader
from pathlib import Path


# =========================
# 1. Configuration
# =========================

class Config:
    # Model configuration
    block_size = 128
    n_layer = 8 #12
    n_head = 8
    n_embd = 512 #768
    dropout = 0.1

    # Training configuration
    batch_size = 16 #32
    learning_rate = 3e-4
    max_iters = 6000

    # Evaluation and generation
    eval_interval = 100 #500
    eval_iters = 5
    max_new_tokens = 1000


config = Config()


# =========================
# 2. Load dataset
# =========================

data_path = Path("input.txt")
text = data_path.read_text(encoding="utf-8")


# =========================
# 3. Character-level dataset
# =========================

# Build one global vocabulary
chars = sorted(list(set(text)))

stoi = {ch: i for i, ch in enumerate(chars)}
itos = {i: ch for ch, i in stoi.items()}

vocab_size = len(chars)

# Encode the entire text
encoded = torch.tensor(
    [stoi[ch] for ch in text],
    dtype=torch.long
)

# Split text into non-overlapping blocks
block_len = config.block_size + 1

n_blocks = len(encoded) // block_len

blocks = encoded[:n_blocks * block_len].view(
    n_blocks, block_len
)

# Randomly split blocks
torch.manual_seed(42)

perm = torch.randperm(n_blocks)

n_train = int(0.9 * n_blocks) # use 90% dataset as train_dataset

train_blocks = blocks[perm[:n_train]]
val_blocks = blocks[perm[n_train:]]


class CharDataset(Dataset):

    def __init__(self, blocks):
        self.blocks = blocks

    def get_vocab_size(self):
        return vocab_size

    def __len__(self):
        return len(self.blocks)

    def __getitem__(self, idx):

        chunk = self.blocks[idx]

        x = chunk[:-1]
        y = chunk[1:]

        return x, y


# =========================
# 4. Initialize datasets
# =========================

train_dataset = CharDataset(train_blocks)
val_dataset = CharDataset(val_blocks)

# Attach the shared vocabulary
train_dataset.stoi = stoi
train_dataset.itos = itos

val_dataset.stoi = stoi
val_dataset.itos = itos

print("Vocabulary size:", vocab_size)
print("Training samples:", len(train_dataset))
print("Validation samples:", len(val_dataset))

x, y = train_dataset[0]
```

```python
# =========================
# 5. Causal Self-Attention
# =========================

class CausalSelfAttention(nn.Module):

    def __init__(self, config):
        super().__init__()

        self.n_head = config.n_head
        self.n_embd = config.n_embd
        self.head_dim = config.n_embd // config.n_head

        # Linear projections for Query, Key, and Value
        self.key = nn.Linear(config.n_embd, config.n_embd)
        self.query = nn.Linear(config.n_embd, config.n_embd)
        self.value = nn.Linear(config.n_embd, config.n_embd)

        # Final projection after combining attention heads
        self.proj = nn.Linear(config.n_embd, config.n_embd)

        # Dropout
        self.attn_dropout = nn.Dropout(config.dropout)
        self.resid_dropout = nn.Dropout(config.dropout)

        # Causal mask: position i cannot attend to positions > i
        self.register_buffer(
            "mask",
            torch.tril(
                torch.ones(
                    config.block_size,
                    config.block_size
                )
            ).view(
                1, 1,
                config.block_size,
                config.block_size
            )
        )

    def forward(self, x):

        B, N, C = x.shape

        # Compute Query, Key, Value
        q = self.query(x)
        k = self.key(x)
        v = self.value(x)

        # Split embedding dimensions into multiple heads
        q = q.view(B, N, self.n_head, self.head_dim).transpose(1, 2)
        k = k.view(B, N, self.n_head, self.head_dim).transpose(1, 2)
        v = v.view(B, N, self.n_head, self.head_dim).transpose(1, 2)

        # Attention scores
        scores = q @ k.transpose(-2, -1)
        scores = scores / (self.head_dim ** 0.5)

        # Apply causal mask
        scores = scores.masked_fill(
            self.mask[:, :, :N, :N] == 0,
            float("-inf")
        )

        # Convert scores into attention probabilities
        attn = F.softmax(scores, dim=-1)
        attn = self.attn_dropout(attn)

        # Weighted sum of Value vectors
        y = attn @ v

        # Combine attention heads
        y = y.transpose(1, 2).contiguous().view(B, N, C)

        # Output projection and dropout
        y = self.proj(y)
        y = self.resid_dropout(y)

        return y
```

```python
# =========================
# 6. Test Causal Self-Attention
# =========================

attention = CausalSelfAttention(config)

# Simulate a batch of token representations
dummy_x = torch.randn(
    2,
    config.block_size,
    config.n_embd
)

attention_output = attention(dummy_x)

print("Input shape:", dummy_x.shape)
print("Output shape:", attention_output.shape)
print("Attention module test passed!")
```
